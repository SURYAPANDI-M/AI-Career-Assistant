```python
import pandas as pd
import numpy as np

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

import joblib
import os
```

```python
!mkdir -p data/raw
!mkdir -p data/processed
!mkdir -p reports
!mkdir -p models
```

```python
jobs_data = {
    "Job_Title": [
        "Data Scientist",
        "Machine Learning Engineer",
        "AI Engineer",
        "Data Analyst",
        "Python Developer",
        "NLP Engineer",
        "Software Engineer"
    ],

    "Skills": [
        "python machine learning statistics sql pandas numpy",
        "python tensorflow pytorch machine learning deep learning aws",
        "python artificial intelligence deep learning tensorflow nlp",
        "sql excel python power bi data visualization",
        "python django flask api sql backend",
        "python nlp transformers bert deep learning",
        "python java c++ algorithms software development"
    ]
}


jobs = pd.DataFrame(jobs_data)

jobs.head()
```

```python
jobs.to_csv(
    "data/raw/jobs.csv",
    index=False
)

print("jobs.csv created successfully")
```

```python
jobs.info()
```

```python
jobs.isnull().sum()
```

```python
import re


def clean_text(text):

    text = text.lower()

    text = re.sub(
        r'[^a-zA-Z\s]',
        '',
        text
    )

    text = re.sub(
        r'\s+',
        ' ',
        text
    )

    return text.strip()
```

```python
jobs["Clean_Skills"] = jobs["Skills"].apply(clean_text)

jobs.head()
```

```python
jobs.to_csv(
    "data/processed/clean_jobs.csv",
    index=False
)

print("clean_jobs.csv saved")
```

```python
vectorizer = TfidfVectorizer()
```

```python
job_vectors = vectorizer.fit_transform(
    jobs["Clean_Skills"]
)
```

```python
job_vectors.shape
```

```python
joblib.dump(
    vectorizer,
    "models/job_vectorizer.pkl"
)

print("Vectorizer saved")
```

```python
def recommend_jobs(resume_text, top_n=5):

    resume_text = clean_text(resume_text)


    resume_vector = vectorizer.transform(
        [resume_text]
    )


    similarity_scores = cosine_similarity(
        resume_vector,
        job_vectors
    )[0]


    top_indexes = similarity_scores.argsort()[::-1][:top_n]


    results = jobs.iloc[top_indexes][
        ["Job_Title","Skills"]
    ].copy()


    results["Match_Score"] = (
        similarity_scores[top_indexes] * 100
    ).round(2)


    return results
```

```python
resume = """
python machine learning sql tensorflow
pandas numpy deep learning
"""
```

```python
recommend_jobs(resume)
```

```python
result = recommend_jobs(resume)


result.to_csv(
    "reports/job_recommendations.csv",
    index=False
)


print(
    "Recommendation report saved"
)
```

```python
import json
import re

file_path = "notebooks/08_Job_Recommendation_System.ipynb"

with open(file_path, "r", encoding="utf-8") as f:
    notebook = json.load(f)

# GitHub token pattern detection
token_regex = re.compile(r"(ghp_[A-Za-z0-9]{20,}|github_pat_[A-Za-z0-9_]{20,})")

found = False

for i, cell in enumerate(notebook["cells"]):
    content = "".join(cell.get("source", []))

    if token_regex.search(content):
        print("Possible token found in code cell:", i)
        found = True

    for j, output in enumerate(cell.get("outputs", [])):
        output_text = json.dumps(output)
        if token_regex.search(output_text):
            print("Possible token found in output:", i, "output", j)
            found = True

if not found:
    print("No GitHub token found in notebook cells or outputs")
```
