```python
# ==========================================
# Phase 7 - Resume Scoring System
# ==========================================

import os
import joblib
import pandas as pd
import numpy as np

from sklearn.metrics.pairwise import cosine_similarity


project_path = "/content/drive/MyDrive/AI-Career-Assistant"

model_path = os.path.join(
    project_path,
    "models"
)

data_path = os.path.join(
    project_path,
    "data"
)

reports_path = os.path.join(
    project_path,
    "reports"
)

os.makedirs(
    reports_path,
    exist_ok=True
)

print("Phase 7 environment ready")
```

```python
tfidf = joblib.load(
    os.path.join(
        model_path,
        "tfidf_vectorizer.pkl"
    )
)

print("TF-IDF model loaded")
```

```python
skills_file = os.path.join(
    data_path,
    "raw/skills.txt"
)


with open(skills_file,"r") as f:
    skills = [
        x.strip().lower()
        for x in f.readlines()
    ]


print("Total skills:", len(skills))
print(skills[:10])
```

```python
def extract_skills(text):

    text = text.lower()

    found = []

    for skill in skills:

        if skill in text:
            found.append(skill)


    return sorted(
        list(set(found))
    )
```

```python
extract_skills(
"""
Python
Machine Learning
TensorFlow
SQL
"""
)
```

```python
def calculate_text_similarity(
    resume,
    job_description
):

    resume_vector = tfidf.transform(
        [resume]
    )


    job_vector = tfidf.transform(
        [job_description]
    )


    similarity = cosine_similarity(
        resume_vector,
        job_vector
    )[0][0]


    return round(
        similarity * 100,
        2
    )
```

```python
def calculate_skill_match(
    resume,
    job_description
):

    resume_skills = set(
        extract_skills(resume)
    )


    job_skills = set(
        extract_skills(job_description)
    )


    if len(job_skills)==0:
        return 0,[],[]


    matched = sorted(
        list(
            resume_skills.intersection(
                job_skills
            )
        )
    )


    missing = sorted(
        list(
            job_skills.difference(
                resume_skills
            )
        )
    )


    score = (
        len(matched)
        /
        len(job_skills)
    )*100


    return (
        round(score,2),
        matched,
        missing
    )
```

```python
def score_resume(
    resume,
    job_description
):

    text_score = calculate_text_similarity(
        resume,
        job_description
    )

    skill_score, matched, missing = calculate_skill_match(
        resume,
        job_description
    )

    final_score = (
        0.6 * skill_score
        +
        0.4 * text_score
    )

    return {

        "Resume Score (%)":
            round(float(final_score), 2),

        "Skill Match (%)":
            round(float(skill_score), 2),

        "Text Similarity (%)":
            round(float(text_score), 2),

        "Matched Skills":
            matched,

        "Missing Skills":
            missing
    }
```

```python
resume = """

Python Developer

Machine Learning

TensorFlow

Pandas

NumPy

SQL

Scikit Learn

Data Analysis

"""
```

```python
job_description = """

Data Scientist

Required Skills:

Python

Machine Learning

SQL

TensorFlow

AWS

Docker

Kubernetes

Statistics

Pandas

"""
```

```python
result = score_resume(
    resume,
    job_description
)

result
```

```python
score_report = pd.DataFrame(
    [result]
)

score_report
```

```python
score_report.to_csv(
    os.path.join(
        reports_path,
        "resume_score_report.csv"
    ),
    index=False
)

print("Resume score report saved")
```

```python
def compare_jobs(resume, jobs):

    results = []

    for role, description in jobs.items():

        score = score_resume(
            resume,
            description
        )

        results.append(
            {
                "Job Role": role,
                "Score": score["Resume Score (%)"]
            }
        )

    df = pd.DataFrame(results)

    # Sort scores
    df = df.sort_values(
        by="Score",
        ascending=False
    ).reset_index(drop=True)

    # Round and convert to display format
    df["Score"] = df["Score"].apply(lambda x: f"{x:.2f}")

    # Add rank
    df.insert(
        0,
        "Rank",
        range(1, len(df) + 1)
    )

    return df
```

```python
jobs = {

"Data Scientist":
"""
Python
SQL
Machine Learning
TensorFlow
Statistics
AWS
""",


"ML Engineer":
"""
Python
Deep Learning
PyTorch
Docker
Kubernetes
AWS
""",


"Backend Developer":
"""
Java
Spring Boot
API
SQL
Database
"""

}
```

```python
result = compare_jobs(resume, jobs)

result.style.hide(axis="index")
```
