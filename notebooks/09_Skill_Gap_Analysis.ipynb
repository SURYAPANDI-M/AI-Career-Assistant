```python
def extract_skills(text):

    skills_list = [
        "python",
        "sql",
        "machine learning",
        "deep learning",
        "tensorflow",
        "pytorch",
        "aws",
        "docker",
        "kubernetes",
        "nlp",
        "pandas",
        "numpy",
        "java",
        "c++",
        "excel",
        "power bi"
    ]


    found_skills = []


    text = text.lower()


    for skill in skills_list:

        if skill in text:
            found_skills.append(skill)


    return found_skills
```

```python
resume_text = """
Python developer with SQL,
TensorFlow and Machine Learning experience
"""

extract_skills(resume_text)
```

```python
job_description = jobs.iloc[0]["Skills"]

job_skills = extract_skills(
    job_description
)

job_skills
```

```python
def find_missing_skills(candidate_skills, job_skills):
    candidate_skills = {
        skill.lower().strip()
        for skill in candidate_skills
    }

    job_skills = {
        skill.lower().strip()
        for skill in job_skills
    }

    return sorted(list(job_skills - candidate_skills))


candidate_skills = [
    "python",
    "sql",
    "pandas",
    "numpy"
]

job_skills = [
    "python",
    "sql",
    "machine learning",
    "tensorflow",
    "aws"
]


missing_skills = find_missing_skills(
    candidate_skills,
    job_skills
)

missing_skills
```

```python
def skill_gap_analysis(
    resume_text,
    job_text
):

    candidate = extract_skills(
        resume_text
    )

    required = extract_skills(
        job_text
    )


    missing = list(
        set(required)-set(candidate)
    )


    return {
        "Candidate Skills": candidate,
        "Required Skills": required,
        "Missing Skills": missing
    }
```

```python
def skill_gap_analysis(
    resume_text,
    job_text
):

    candidate = extract_skills(
        resume_text
    )

    required = extract_skills(
        job_text
    )


    missing = list(
        set(required)-set(candidate)
    )


    return {
        "Candidate Skills": candidate,
        "Required Skills": required,
        "Missing Skills": missing
    }
```

```python
result = skill_gap_analysis(
    """
    python sql machine learning pandas
    """,

    """
    python sql machine learning aws docker kubernetes
    """
)


result
```

```python
report = pd.DataFrame(
    {
        "Missing Skills":
        result["Missing Skills"]
    }
)


report.to_csv(
    "reports/skill_gap_report.csv",
    index=False
)


print(
    "Skill gap report created"
)
```
